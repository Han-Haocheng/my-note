### 6.2.3 更新集合（可变集合）

```python
student1_set = {'张明','赵虎','马汉','李云龙','王义'}
student2_set = {'孙悟空','猪八戒','沙僧'}
print("初始化集合student1_set为：")
print(student1_set)
student1_set.add('孙悟空')
print("使用函数add（）添加元素‘孙悟空’后的集合student1_set为：")
print(student1_set)
student1_set.update(student2_set)
print("使用函数update（）一次性添加多个元素后的集合student1_set为：")
print(student1_set)
```

```python
student_set={'孙悟空','猪八戒','沙僧'}
print("初始化集合student_set为：")
print(student_set)
student_set.remove('孙悟空')
print("使用函数remove（）删除元素‘孙悟空’后的集合student_set为：")
obj = student_set.pop()
print("使用函数pop（）删除一个元素后的集合student_set为：")
print(student_set)
if obj=='猪八戒':
    student_set.discard('沙僧')
    print("使用函数discard（）删除元素‘沙僧’后的集合student_set为：")
    print(student_set)
else:
    student_set.discard('猪八戒')
    print("使用函数discard（）删除元素‘猪八戒’后的集合student_set为：")
    print(student_set)
student_set.clear()
print("使用函数clear（）清空集合student_set后的集合为：")
print(student_set)
```

### 6.2.4 集合操作符

```python
student1_set = {'张明','赵虎','马汉'}
student2_set = {'张明','孙悟空','猪八戒'}
print("初始化集合student1_set为：")
print(student1_set)
print("初始化集合student2_set为：")
print(student2_set)

print("student1_set|student2_set：")
print(student1_set|student2_set)

print("student1_set&student2_set：")
print(student1_set&student2_set)
print("student1_set-student2_set：")
print(student1_set-student2_set)
print("student1_set^student2_set：")
print(student1_set^student2_set)

```

### 6.2.5 常用集合内置函数和方法

```python
student1_set = {'张明','赵虎','马汉'}
student2_set = {'张明','孙悟空','猪八戒'}
print("初始化集合student1_set为：")
print(student1_set)
print("初始化集合student2_set为：")
print(student2_set)

print('student1_set.union(student2_set)')
student1_set.union(student2_set)

print('student1_set.intersection(student2_set)')
print(student1_set.intersection(student2_set))

print('student1_set.difference(student2_set)')
print(student1_set.difference(student2_set))

print('student1_set.symmetric_difference(student2_set)')
print(student1_set.symmetric_difference(student2_set))
```
