## 3.3 列表元素更新
### 3.3.1 使用赋值语句更新列表元素

```python
students = ["张明","赵虎","马汉","李云龙","王义"]
print(students)
students[1]="李勇"
print(students)
students[2:4]=["期中","期末"]
print(students)
```

## 3.4 列表元素的删除
### 3.4.1 使用del语句删除列表元素或列表本身

```python
students = ["张明","赵虎","马汉","李云龙","王义"]
del students[1]
print(students)
students = ["张明","赵虎","马汉","李云龙","王义"]
del students[1:3]
print(students)
students = ["张明","赵虎","马汉","李云龙","王义"]
del students[1:2]
print(students)
students = ["张明","赵虎","马汉","李云龙","王义"]
del students
print(students)
```

### 3.4.2 使用remove方法删除列表元素

```python
students = ["张明","赵虎","马汉","李云龙","王义","赵虎"]
print(students)
students.remove("赵虎")
print(students)
students.remove("王五")
```

### 3.4.3 使用pop方法移除列表元素

```python
students = ["张明","赵虎","马汉","李云龙","王义","赵虎"]
print(students)
obj = students.pop()
print(students)
print(obj)
```

```python
students = ["张明","赵虎","马汉","李云龙","王义","赵虎"]
print(students)
obj = students.pop(3)
print(students)
print(obj)
students = ["张明","赵虎","马汉","李云龙","王义","赵虎"]
print(students)
obj = students.pop(6)
```

## 3.5 列表的排序和比较
### 3.5.1 使用列表内置的sort方法进行排序

```python
cars = ["bmw","audi","toyota","subaru"]
print(cars)
cars.sort()
print(cars)
cars.sort(reverse=True)
print(cars)
```

### 3.5.2 使用通用函数sorted()方法进行排序

```python
cars = ["bmw","audi","toyota","subaru"]
print("sorted list:")
print(sorted(cars))
print("sorted list(reverse):")
print(sorted(cars,reverse=True))
print("origin list:")
print(cars)
```

```python
cars = ["bmw","audi","toyota","subaru"]
print(cars)
cars.reverse()
print(cars)
```

## 3.6 求列表的最大值和最小值

```python
cars = ["bmw","audi","toyota","subaru"]
print("最大值："+max(cars)+"\t\t最小值："+min(cars))
sealdata =[128.92,65.90,13809,79.689,0.23,127.00]
print("最大值：%.2f\t\t最小值：%.2f"%(max(sealdata),min(sealdata)))
```
