## 5.5 对数字列表执行简单的统计

```python
lst=[20,16,3,13,14,6,6,6,17,17]
print("min = ",min(lst))
print("max = ",max(lst))
print("sum = ",sum(lst))
```

```python
lst = [20,16,3,13,14,6,6,6,17,17]
count = [0]*20
for i in lst:
    if count[i-1]!=0:
        continue
    count[i-1] = lst.count(i)
for j in range(len(lst)):
    print("nums:{:<2}    dig:{}".format(lst[j],count[j-1]))
```

## 进阶B python3 浅拷贝与深拷贝的区别
### B.1 浅拷贝

```python
import copy
num1 = 17
mum2 = copy.copy(num1)
print("num1=",str(id(num1)))
print("mum2=",str(id(mum2)))
print("-------------------")
str1 = 'hello'
str2 = copy.copy(str1)
print("str1=",str(id(str1)))
print("str2=",str(id(str2)))
print("-------------------")
tup1 = (18,"tom")
tup2 = copy.copy(tup1)
print("tup1=",str(id(tup1)))
print("tup2=",str(id(tup2)))
```

```python
list1=[11,12]
list2=copy.copy(list1)
list3=list1
print("list1=",str(id(list1)))
print("list2=",str(id(list2)))
print("list3=",str(id(list3)))
print("the first elements of list1:",str(id(list1[0])))
print("the first elements of list2:",str(id(list2[0])))
set1={"AA","BB"}
set2 = copy.copy(set1)
print("--------------------")
print("set1=",str(id(set1)))
print("set2=",str(id(set2)))
```

```python
l1 = [11,12];l2 = [21,22]; num = 555
allOne = [l1,l2,num]
allOne2 = copy.copy(allOne)
l1[0] = 16;allOne[2]=666; num = 777
print(allOne)
print(allOne2)
print("id allOne:",str(id(allOne)))
print("id allOne[0]:",str(id(allOne[0])))
print("id allOne[1]:",str(id(allOne[1])))
print("id allOne[2]:",str(id(allOne[2])))
print("id allOne2:",str(id(allOne2)))
print("id allOne2[0]:",str(id(allOne2[0])))
print("id allOne2[1]:",str(id(allOne2[1])))
print("id allOne2[2]:",str(id(allOne2[2])))
```

```python
num = 123
print(str(id(num)))
num=666
print(str(id(num)))
```

### B.2 深拷贝

```python
import copy
num1 = 17
mum2 = copy.deepcopy(num1)
print("num1=",str(id(num1)))
print("mum2=",str(id(mum2)))
print("-------------------")
str1 = 'hello'
str2 = copy.deepcopy(str1)
print("str1=",str(id(str1)))
print("str2=",str(id(str2)))
print("-------------------")
tup1 = (18,"tom")
tup2 = copy.deepcopy(tup1)
print("tup1=",str(id(tup1)))
print("tup2=",str(id(tup2)))
```

```python
list1=[11,12]
list2=copy.deepcopy(list1)
list3=list1
print("list1=",str(id(list1)))
print("list2=",str(id(list2)))
print("list3=",str(id(list3)))
print("the first elements of list1:",str(id(list1[0])))
print("the first elements of list2:",str(id(list2[0])))
set1={"AA","BB"}
set2 = copy.deepcopy(set1)
print("--------------------")
print("set1=",str(id(set1)))
print("set2=",str(id(set2)))
```

```python
l1 = [11,12];l2 = [21,22]; num = 555
allOne = [l1,l2,num]
allOne2 = copy.deepcopy(allOne)
allOne[1] = [113,113];allOne2[2]=[277,228]
print(allOne);print(allOne2)
print("id allOne:",str(id(allOne)))
print("id allOne[0]:",str(id(allOne[0])))
print("id allOne[1]:",str(id(allOne[1])))
print("id allOne[2]:",str(id(allOne[2])))
print("id allOne2:",str(id(allOne2)))
print("id allOne2[0]:",str(id(allOne2[0])))
print("id allOne2[1]:",str(id(allOne2[1])))
print("id allOne2[2]:",str(id(allOne2[2])))
```

### B.3 其他拷贝方式

```python
l1 = [11,12];l2 = [21,22]; num = 555
orgi = [l1,l2,num]
nList = orgi[:]
print("orgi:",str(id(orgi)))
print("orgi[0]:",str(id(orgi[0])))
print("orgi[1]:",str(id(orgi[1])))
print("orgi[2]:",str(id(orgi[2])))
print("*"*30)
print("nList:",str(id(nList)))
print("nList[0]:",str(id(nList[0])))
print("nList[1]:",str(id(nList[1])))
print("nList[2]:",str(id(nList[2])))
```
