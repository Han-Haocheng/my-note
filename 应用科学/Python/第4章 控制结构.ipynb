## 4.3 分支结构
### 4.3.1 单分支结构：if语句

```python
username = input('请输入用户名：')
if username == 'admin':
    print('提示：'+username+'是管理员用户')
```

### 4.3.2 二分支结构：if-else语句

```python
username = input('请输入用户名：')
if username == 'admin':
    print('提示：'+username+'是管理员用户！')
else:
    print('提示：'+username+'不是管理员用户！')
```

```python
x=float(input('x='))
if x>1:
    y=3*x-5
else:
    if x>=-1:
        y=x+2
    else:
        y=5*x+3
print('f(%.2f)=%.2f'%(x,y))
```

```python
grade = int(input("请输入成绩grade："))
if grade >=90:
    print('A')
else:
    if grade >=80:
        print('B')
    else:
        if grade >=70:
            print('C')
        else:
            if grade >=60:
                print('D')
            else:
                print('E')
```

### 4.3.3 多分支结构：if-elif-else语句

```python
x=float(input('x='))
if x>1:
    y=3*x-5
elif x>=-1:
    y=x+2
else:
    y=5*x+3
print('f(%.2f)=%.2f'%(x,y))
```

```python
grade = int(input("请输入成绩grade："))
if grade >=90:
    print('A')
elif grade >=80:
    print('B')
elif grade >=70:
    print('C')
elif grade >=60:
    print('D')
else:
    print('E')
```

## 4.4 循环结构
### 4.4.4 遍历循环：for语句

```python
print("输出name中的每一个字符")
name = '张三'
for ch in name:
    print(ch)
```

```python
fruits=['banana','apple','mango']
for fruit in fruits:
    print('当前水果：'+fruit)
```

```python
fruits=['banana','apple','mango']
for index in range(len(fruits)):
    print('当前水果：'+fruits[index])
```

```python
for i in range(5):
    print(i)
else:
    print("hello?")
```

```python
for r in range(3):
    for c in range(5):
        print('*',end='')
    print()
```

```python
# 输出九九乘法表
str_ =''
for i in range(1,10):
    for j in range(1,i+1):
        str_+='%d*%d=%d\t'%(i,j,i*j)
    str_+='\n'
print(str_)
```

### 4.4.2 条件循环：while语句

```python
num = 1
while num <= 10:
    print('num=',num)
    num += 1
print("循环结束！")
```

```python
var = 1
while var ==1:
    num = input('请输入一个数字：')
    print('你输入的是：',num)
print ("good bye")
```

```python
count = 0
while count < 5:
    print(count,' is less than 5')
    count += 1
else:
    print(count,' is not less than 5')
```

```python
i=1
while i<=5:
    j=1
    while j<=i:
        print('*',end='')
        j+=1
    print('')
    i+=1
```

```python
row =1
while row <=9:
    col =1
    while col <=row:
        print('%d*%d=%d\t'%(col,row,col*row),end='')
        col+=1
    print("")
    row +=1
```

### 4.4.3 循环保留字：break和continue

```python
i = 1
while 1:
    print(i)
    i += 1
    if(i>10):
        break
```

```python
for i in range(0,5):
    if i >3:
        print(i)
        break
else:
    print('hello world')
```

```python
for num in range(10,21):
    for i in range(2,num):
        if num%i ==0:
            j=num/i
            print('%d 等于 %d * %d'%(num,i,j))
            break
    else:
        print(num,'是一个质数')
```

```python
num = int(input("请输入一个大于1的整数："))
count = num//2
while count > 1:
    if num % count ==0:
        print("%d 的最大公约数是%d"%(num,count))
        break
    count-=1
else:
    print('%d是一个质数'%num)
```

```python
i =1
while i<=10:
    i+=1;
    if (i%2) ==1:
        continue
    print(i)
```
